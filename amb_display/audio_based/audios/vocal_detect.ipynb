```python
import pyaudio
import numpy as np
import torch
import librosa
import librosa.display
import matplotlib.pyplot as plt
from silero_vad import get_speech_timestamps

# Initialize Silero VAD model
model, utils = torch.hub.load('snakers4/silero-vad', 'silero_vad', force_reload=True)


# PyAudio setup
CHUNK = 1024
RATE = 16000  # Sample rate
FORMAT = pyaudio.paInt16
CHANNELS = 1

p = pyaudio.PyAudio()
stream = p.open(format=FORMAT, channels=CHANNELS, rate=RATE, input=True, frames_per_buffer=CHUNK,)

def detect_vocals(audio_chunk):
    """Check if vocals are detected."""
    audio_np = np.frombuffer(audio_chunk, dtype=np.int16).astype(np.float32) / 32768.0
    speech_timestamps = get_speech_timestamps(torch.tensor(audio_np), model, sampling_rate=RATE)
    return bool(speech_timestamps)

def analyze_frequency(audio_chunk):
    """Compute the frequency spectrum of the audio chunk."""
    # Convert audio to floating point
    audio_np = np.frombuffer(audio_chunk, dtype=np.int16).astype(np.float32) / 32768.0

    
    # Compute STFT (Short-Time Fourier Transform)
    stft_result = librosa.stft(audio_np, n_fft=1024, hop_length=512)
    magnitude = np.abs(stft_result)

    # Get frequency axis in Hz
    freqs = librosa.fft_frequencies(sr=RATE, n_fft=1024)

    # Plot the frequency spectrum
    plt.figure(figsize=(10, 4))
    plt.semilogy(freqs, np.mean(magnitude, axis=1))  # Average magnitude over frames
    plt.xlabel("Frequency (Hz)")
    plt.ylabel("Magnitude")
    plt.title("Frequency Spectrum of Detected Vocals")
    plt.grid()
    plt.show()

print("Listening for vocals...")

try:
    while True:
        audio_chunk = stream.read(CHUNK, exception_on_overflow=False)
        
        if detect_vocals(audio_chunk):
            print("Vocals detected!")
            analyze_frequency(audio_chunk)
        else:
            print("No vocals detected")

except KeyboardInterrupt:
    print("\nStopping...")
finally:
    stream.stop_stream()
    stream.close()
    p.terminate()
```
